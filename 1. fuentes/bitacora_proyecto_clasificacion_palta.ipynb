## 10. Riesgos metodológicos

### 10.1 Data leakage

Debe revisarse especialmente si existen imágenes repetidas entre:

```text
train ↔ validation
train ↔ test
validation ↔ test
```

Se recomienda:

1. Hash convencional para detectar duplicados exactos.
2. Perceptual hash para detectar imágenes visualmente casi idénticas.

Un mismo contenido presente en train y test puede generar métricas artificialmente altas, ya que el modelo estaría siendo evaluado con información que ya vio durante el entrenamiento.

### 10.2 Desbalance

Se debe cuantificar el desbalance entre clases.

Luego, dependiendo del resultado, se pueden evaluar:

- `class_weights`;
- data augmentation;
- oversampling;
- undersampling.

Estas estrategias no deben aplicarse automáticamente durante el EDA.

### 10.3 Diferencias de distribución

Como el dataset es estático, no se debe afirmar automáticamente que existe *data drift temporal*.

Sí se pueden analizar posibles diferencias entre train, validation y test en:

- resolución;
- brillo;
- RGB;
- relación de aspecto;
- proporción de clases.

## 15. Plantilla inicial para inspeccionar un dataset de imágenes

Se conserva esta consulta sencilla como antecedente del inventario. El EDA completo está en `4. EDA/01_EDA.ipynb`.

El recuento está desactivado por defecto. Para realizarlo, el autor puede cambiar `RECONTAR_IMAGENES` a `True` después de verificar cuál de las dos ubicaciones relativas corresponde al dataset original. No se copian ni modifican imágenes.

````python
from collections import Counter

RECONTAR_IMAGENES = False
EXTENSIONES_IMAGEN = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".webp"}

rutas_candidatas = (
    raiz_proyecto / "DATASET_PALTA",
    raiz_proyecto.parent / "DATASET_PALTA",
)
if RECONTAR_IMAGENES:
    rutas_disponibles = [ruta for ruta in rutas_candidatas if ruta.is_dir()]
    if len(rutas_disponibles) != 1:
        print("Debe existir una única ubicación seleccionada para DATASET_PALTA.")
    else:
        ruta_datos = rutas_disponibles[0]
        archivos_imagen = [
            archivo
            for archivo in ruta_datos.rglob("*")
            if archivo.is_file() and archivo.suffix.lower() in EXTENSIONES_IMAGEN
        ]
        print(f"Imágenes encontradas: {len(archivos_imagen)}")
        formatos = Counter(archivo.suffix.lower() for archivo in archivos_imagen)
        for formato, cantidad in formatos.most_common():
            print(f"{formato}: {cantidad}")
else:
    print("Recuento desactivado. Consulte los reportes de EDA conservados.")
````
