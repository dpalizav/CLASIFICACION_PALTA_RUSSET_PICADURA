# Matriz de consistencia

## Diferenciación de defectos superficiales en palta mediante visión artificial y aprendizaje profundo


## 1. Alcance y unidad de análisis

El proyecto estudia la diferenciación de dos defectos superficiales visibles en palta Hass: PICADURA y RUSSET. Se utiliza una arquitectura ResNet50 entrenada con imágenes RGB multivista.

La unidad experimental es la palta individual, representada por un archivo TIFF. La imagen contiene tres franjas horizontales correspondientes a tres cámaras; cada franja puede incluir apariciones de la misma fruta durante su rotación. Las vistas de una palta no deben distribuirse entre conjuntos diferentes.

El inventario original comprende 829 archivos. Tras excluir cuatro registros por los criterios documentados en la auditoría, se consideran 825 muestras elegibles: 572 de train, 127 de validation y 126 de TEST. El conjunto TEST se evaluó una única vez y no se utilizará para ajustar las perturbaciones del sprint 7.

## 2. Matriz de consistencia

La matriz vincula el problema de investigación con los objetivos, hipótesis, variables e indicadores. Las fichas siguientes reproducen la matriz consolidada para la semana 7. Al final del notebook hay una celda para mostrarla también como tabla de siete columnas.

### 2.1. Nivel general

Problema: ¿En qué medida una metodología basada en ResNet50 permite diferenciar los defectos superficiales PICADURA y RUSSET a partir de imágenes RGB multivista de palta Hass?

Objetivo: Evaluar el desempeño de una metodología de visión artificial y aprendizaje profundo basada en ResNet50 para diferenciar defectos superficiales PICADURA y RUSSET en imágenes multivista de palta Hass.

Hipótesis: La metodología ResNet50 con ajuste fino y aumentos fotométricos presenta un mayor Macro-F1 en validation que la configuración base B0, bajo las particiones y protocolo de este estudio.

Variable independiente: Configuración de la metodología de clasificación (ResNet50 base, balanceo, resolución, ajuste fino y aumentos fotométricos).

Variable dependiente: Desempeño de clasificación de defectos PICADURA y RUSSET.

Indicadores: Macro-F1 (principal); accuracy, recall, precisión por clase y matriz de confusión (complementarios).

### 2.2. Nivel específico 1

Problema: ¿Cómo varía el desempeño de ResNet50 con diferentes configuraciones de resolución, balanceo, ajuste fino y aumentos fotométricos?

Objetivo: Comparar el desempeño de las configuraciones B0, V1, D1, D2, FT_B0, FT_D1 y FT_D1_PHOTO sobre validation.

Hipótesis: La configuración FT_D1_PHOTO alcanza un Macro-F1 superior al de B0 sobre la misma partición validation.

Variable independiente: Configuración experimental de ResNet50 (categorías B0, V1, D1, D2, FT_B0, FT_D1, FT_D1_PHOTO).

Variable dependiente: Desempeño de clasificación en validation.

Indicadores: Macro-F1, accuracy, recall RUSSET, precisión RUSSET. Evidencia disponible: Macro-F1 B0 0.6604 y FT_D1_PHOTO 0.7876.

### 2.3. Nivel específico 2

Problema: ¿Qué sensibilidad presentan las predicciones de ResNet50 al alterar la información de cada cámara y las condiciones de brillo y contraste?

Objetivo: Cuantificar la sensibilidad del modelo seleccionado a la ocultación/desenfoque de cada franja y a variaciones fotométricas moderadas, usando validation.

Hipótesis: La alteración de franjas individuales y de brillo/contraste produce cambios diferenciados de predicción y Macro-F1 respecto de la imagen completa.

Variable independiente: Condición de entrada: cámara 1, 2 o 3 intervenida; tipo de perturbación; factor de brillo/contraste (0.95, 1.05).

Variable dependiente: Sensibilidad del clasificador en validation.

Indicadores: Caída de Macro-F1; cambio promedio absoluto de probabilidad RUSSET; número de etiquetas que cambian; IC bootstrap exploratorio. Pendiente de ejecutar.

### 2.4. Nivel específico 3

Problema: ¿Qué capacidad de generalización presenta el modelo seleccionado sobre imágenes reservadas no usadas para entrenar ni seleccionar configuraciones?

Objetivo: Evaluar una sola vez el modelo congelado FT_D1_PHOTO sobre las imágenes reservadas de TEST.

Hipótesis: El modelo seleccionado es capaz de identificar ambas clases en TEST y supera descriptivamente al predictor de clase mayoritaria en Macro-F1.

Variable independiente: Modelo congelado seleccionado con validation: FT_D1_PHOTO; comparador descriptivo: predictor de la clase mayoritaria.

Variable dependiente: Desempeño de clasificación en TEST.

Indicadores: Accuracy 0.9444; Macro-F1 0.9019; recall RUSSET 0.9000; matriz [[101, 5], [2, 18]], n=126.

## 3. Operacionalización de variables

Las variables principales son la configuración metodológica del clasificador (independiente) y el desempeño de la clasificación (dependiente). Para el sprint 7 se incorporan dimensiones de sensibilidad visual: alteración de una de las franjas, variación moderada de iluminación y análisis cualitativo de activaciones.

La ocultación o el desenfoque de una cámara se interpretan como perturbaciones experimentales de entrada. Una disminución de Macro-F1 no demuestra por sí sola causalidad ni calidad física inferior de esa cámara; indica sensibilidad del clasificador bajo esa intervención.

### 3.1. Configuración ResNet50

Concepto: Variable independiente global

Definición operativa: Condición de entrenamiento y preparación de imagen.

Valores o escala: Nominal: B0, V1, D1, D2, FT_B0, FT_D1, FT_D1_PHOTO

Indicador: Identificador de experimento

Fuente de datos: `Resultados de validation`

Estado: Verificado

### 3.2. Desempeño de clasificación

Concepto: Variable dependiente global

Definición operativa: Comparación entre etiquetas verdaderas y predicciones del modelo.

Valores o escala: Escala de 0 a 1

Indicador: Macro-F1, accuracy, recall y precisión por clase

Fuente de datos: `metricas_validation.json / metricas_test.json`

Estado: Verificado

### 3.3. Información de cada cámara

Concepto: Dimensión visual 1

Definición operativa: Una de las tres bandas horizontales de un TIFF multivista se altera en memoria.

Valores o escala: Cámara 1, 2, 3; ocultar o desenfocar

Indicador: Diferencia de Macro-F1 y cambios de etiqueta vs referencia

Fuente de datos: `04_Importancia_Variables.ipynb`

Estado: Pendiente de ejecución

### 3.4. Brillo y contraste

Concepto: Dimensión visual 2

Definición operativa: Se aplica una perturbación determinista leve sobre el TIFF antes de redimensionarlo.

Valores o escala: Brillo/contraste: 0.95 y 1.05

Indicador: Diferencia de Macro-F1 y cambio promedio de p(RUSSET)

Fuente de datos: `04_Importancia_Variables.ipynb`

Estado: Pendiente de ejecución

### 3.5. Grad-CAM

Concepto: Dimensión interpretativa

Definición operativa: Activaciones de la última etapa convolucional para la clase predicha.

Valores o escala: Mapa de calor cualitativo

Indicador: Regiones de activación observadas

Fuente de datos: `04_Importancia_Variables.ipynb`

Estado: Opcional

### 3.6. Palta individual

Concepto: Unidad de análisis

Definición operativa: Cada archivo TIFF corresponde a un fruto con tres vistas de cámara y una etiqueta.

Valores o escala: 825 elegibles: 572 train, 127 validation, 126 TEST

Indicador: Número de TIFF por partición y clase

Fuente de datos: `EDA y auditoría del protocolo`

Estado: Verificado

## 5. Precauciones metodológicas

- Selección del modelo: se utilizó validation para comparar configuraciones. La comparación B0 frente a FT_D1_PHOTO corresponde a esa partición; no se ha reentrenado el modelo para el sprint 7.
- Reproducibilidad: al reconstruir FT_D1_PHOTO desde el checkpoint, 126 de las 127 predicciones de validation coincidieron con el registro histórico. La diferencia máxima absoluta de probabilidad fue 0.004967. El archivo discrepante estaba cerca del umbral de 0.5. No se presenta la reproducción como exacta.
- Prueba final: el modelo congelado se evaluó una sola vez sobre TEST. Se obtuvo Macro-F1 0.9019, accuracy 0.9444 y recall RUSSET 0.9000. No corresponde reutilizar TEST para elegir nuevas perturbaciones.
- Sensibilidad visual: las métricas de ocultación, desenfoque y variación fotométrica están pendientes de ejecución local en validation. No se asignan importancias antes de calcularlas.
- Hipótesis: las afirmaciones de la matriz se interpretan en el marco de la muestra disponible. La comparación con un clasificador de clase mayoritaria es una referencia descriptiva, no un ensayo confirmatorio preregistrado.
- Alcance: los resultados no prueban el mismo rendimiento para distintas campañas, cámaras o instalaciones agroindustriales.
