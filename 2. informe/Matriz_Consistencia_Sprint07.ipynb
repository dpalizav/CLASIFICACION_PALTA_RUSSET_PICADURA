## 1. Alcance y unidad de análisis

El proyecto analiza la diferenciación de los defectos PICADURA y RUSSET en palta Hass mediante una única arquitectura principal, ResNet50. Cada archivo TIFF contiene tres franjas horizontales de capturas de cámara y constituye la unidad de análisis; las vistas de un mismo TIFF no se distribuyen entre particiones diferentes.

El inventario original comprende 829 TIFF. Se aplicaron cuatro exclusiones lógicas documentadas, por lo que se dispone de 825 archivos elegibles: 572 para entrenamiento, 127 para validación y 126 para TEST reservado. La evaluación final del modelo anterior en TEST se ejecutó una sola vez. No se usará TEST para seleccionar nuevas condiciones ni para buscar hiperparámetros en el sprint 8.

Las métricas de los experimentos históricos se mantienen tal como fueron registradas. La reproducción posterior de la validación del modelo seleccionado fue compatible, pero no exactamente idéntica al archivo histórico (126/127 predicciones coincidentes); por eso se distinguen ambos resultados.

## 2. Planteamiento de consistencia

Los objetivos emplean verbos de medición y declaran los indicadores. Las hipótesis son contrastables con los registros disponibles y las variables describen qué condición se modifica o mantiene fija. No se presenta una decisión obtenida después del experimento como si hubiese sido una hipótesis registrada previamente.


### 2.1. Nivel general

Problema general: ¿Cómo varía la capacidad de ResNet50 para diferenciar PICADURA y RUSSET en imágenes RGB multivista de palta Hass según su configuración de entrenamiento y la integridad de la información visual de entrada?

Objetivo general: Cuantificar el desempeño de ResNet50 y su variación ante configuraciones de entrenamiento e intervenciones visuales, mediante Macro-F1, precisión y recall por clase en particiones reservadas.

Hipótesis general: La configuración ResNet50 con ajuste fino parcial, compensación de RUSSET y aumentos fotométricos presenta un Macro-F1 mayor que el modelo base ResNet50 con extractor congelado en la misma partición de validación.

Variable independiente: Configuración de entrenamiento de ResNet50 y condición visual de la entrada multivista.

Variable dependiente: Desempeño de clasificación de PICADURA y RUSSET.

Indicadores: Macro-F1 (principal); precisión y recall por clase; accuracy y matriz de confusión (complementarios).


### 2.2. Nivel específico 1 · configuración de ResNet50

Problema específico 1: ¿Cuál es la diferencia de Macro-F1 y recall RUSSET entre el modelo base ResNet50 y una configuración con ajuste fino parcial, compensación de RUSSET y aumentos fotométricos?

Objetivo específico 1: Medir las diferencias de Macro-F1 y recall RUSSET entre el modelo base ResNet50 con extractor congelado y ResNet50 con ajuste fino parcial, compensación de RUSSET y aumentos fotométricos, sobre las mismas imágenes de validación.

Hipótesis específico 1: La configuración con ajuste fino parcial, compensación de RUSSET y aumentos fotométricos presenta valores mayores de Macro-F1 y recall RUSSET que el modelo base ResNet50 con extractor congelado en validación.

Variable independiente: Configuración de entrenamiento: modelo base ResNet50 con extractor congelado (B0) frente a ResNet50 con ajuste fino y aumentos fotométricos (FT_D1_PHOTO).

Variable dependiente: Desempeño del clasificador en validación.

Indicadores: Diferencia de Macro-F1 y de recall RUSSET; precisión RUSSET y accuracy como métricas complementarias.


### 2.3. Nivel específico 2 · información visual conservada

Problema específico 2: ¿Cuánto cambian el Macro-F1, el recall RUSSET y las etiquetas predichas al ocultar o desenfocar individualmente las tres franjas de cámara o modificar brillo y contraste?

Objetivo específico 2: Cuantificar las pérdidas y ganancias de Macro-F1, precisión y recall RUSSET y los cambios de etiqueta ante seis intervenciones de cámara y cuatro variaciones de brillo y contraste, para sustentar qué información visual conservar en la entrada del clasificador.

Hipótesis específico 2: La alteración de al menos una franja reduce el Macro-F1 frente a la imagen completa; las perturbaciones fotométricas moderadas (±5 %) producen cambios de menor magnitud que la mayor caída observada por intervención de franja.

Variable independiente: Franja intervenida (1, 2 o 3) y tratamiento (ocultación/desenfoque); brillo y contraste multiplicados por 0.95 o 1.05.

Variable dependiente: Variación del desempeño y de las predicciones del clasificador en validación.

Indicadores: ΔMacro-F1 respecto a la imagen completa, accuracy, precisión y recall RUSSET, etiquetas modificadas, cambio absoluto de p(RUSSET) e intervalos bootstrap exploratorios.


### 2.4. Nivel específico 3 · errores en TEST

Problema específico 3: ¿Cuántos errores PICADURA→RUSSET y RUSSET→PICADURA presenta el modelo seleccionado al clasificar el conjunto TEST reservado?

Objetivo específico 3: Medir la frecuencia y las tasas de falsos positivos y falsos negativos de RUSSET, el Macro-F1 y la exactitud del modelo seleccionado, utilizando las 126 imágenes de TEST reservadas.

Hipótesis específico 3: El modelo seleccionado presenta un Macro-F1 superior al de la regla descriptiva que predice siempre la clase mayoritaria, al clasificar el conjunto TEST reservado.

Variable independiente: Condición de evaluación fija: ResNet50 previamente seleccionado con validación (FT_D1_PHOTO) y conjunto TEST reservado; no se manipula una variable independiente adicional durante esta prueba.

Variable dependiente: Desempeño y errores de clasificación en TEST.

Indicadores: Falsos positivos de RUSSET (PICADURA→RUSSET), falsos negativos de RUSSET (RUSSET→PICADURA), tasas correspondientes, recall y precisión RUSSET, accuracy y Macro-F1.


### 2.5. Decisión sobre conservación o exclusión de información visual

La selección de variables visuales se basa en la variación del Macro-F1, el recall y la precisión RUSSET; no basta con una mejora aislada de accuracy.

| Información | Observación en 127 imágenes de validación | Decisión fundamentada |
|---|---|---|
| Franja 1 | Al ocultarla, Macro-F1 = 0.7354 frente a 0.7781 de referencia; recall RUSSET baja a 0.3810. | Conservarla en la entrada actual. |
| Franja 2 | Ocultarla eleva levemente Macro-F1 a 0.7883, pero desenfocarla lo reduce a 0.6945; el intervalo bootstrap de la diferencia por ocultación incluye 0. | Conservarla provisionalmente; evidencia insuficiente para eliminarla. |
| Franja 3 | Ocultación: Macro-F1 = 0.6066; desenfoque: Macro-F1 = 0.5609. | Conservarla: mayor sensibilidad observada bajo ambas intervenciones. |
| Brillo ×0.95 | Macro-F1 sube a 0.7900, pero recall RUSSET cae de 0.7143 a 0.6667. | No adoptar automáticamente como nuevo preprocesamiento. |
| Contraste ×0.95 | Macro-F1 sube a 0.7974, con recall RUSSET de 0.7143. | Mantener como opción exploratoria para estudio controlado posterior. |

Decisión del sprint 7: se mantienen las tres franjas de cámara y el procesamiento de entrada seleccionado. No se justifica retirar hardware ni modificar la operación de línea industrial a partir de estas perturbaciones exploratorias. Eliminar una franja de un modelo entrenado con tres no equivale a entrenar una arquitectura diseñada para dos cámaras.


## 3. Operacionalización de variables

La configuración del modelo y la condición de entrada son factores independientes estudiados en experimentos distintos. El desempeño es la variable dependiente, medida con resultados de clasificación. En la evaluación TEST el modelo y el protocolo están fijos: no se modifica un factor de entrenamiento.

Los mapas Grad-CAM son una evidencia cualitativa complementaria, no una estimación numérica de importancia de cámara ni una demostración de localización precisa del defecto.


### 3.1. Configuración del clasificador

Concepto: Independiente; entrenamiento

Variable/dimensión: Configuración del clasificador ResNet50

Definición operativa: Se contrasta el modelo base con el modelo de ajuste fino y aumentos fotométricos; los experimentos intermedios se documentan en el historial de entrenamiento.

Valores / escala: Nominal: modelo base (B0) y ajuste fino + compensación RUSSET + fotometría (FT_D1_PHOTO).

Indicador: Código de configuración y parámetros de entrenamiento registrados.

Fuente de datos: 7. resultados/entrenamiento/*/configuracion.json y métricas históricas.

Estado: Ejecutado; antecedentes guardados.


### 3.2. Desempeño de clasificación

Concepto: Dependiente; clasificación

Variable/dimensión: Desempeño del clasificador

Definición operativa: Métricas obtenidas al contrastar etiquetas reales con las predicciones del clasificador bajo una condición definida.

Valores / escala: Macro-F1, accuracy, precisión y recall entre 0 y 1; matriz de confusión en conteos.

Indicador: Macro-F1 principal; precisión, recall, accuracy y confusiones como complementos.

Fuente de datos: Métricas de validation y TEST ya registradas.

Estado: Ejecutado.


### 3.3. Intervenciones sobre franjas

Concepto: Independiente; visión

Variable/dimensión: Información de las franjas de cámara

Definición operativa: Se oculta o desenfoca una de las tres franjas de cada TIFF en memoria, manteniendo las otras dos.

Valores / escala: Nominal: C1, C2, C3 × ocultación/desenfoque; referencia sin intervención.

Indicador: ΔMacro-F1, recall RUSSET, precisión RUSSET, cambios de etiqueta, intervalo bootstrap.

Fuente de datos: 6. entrenamiento/04_Importancia_Variables.ipynb; 7. resultados/importancia_variables/.

Estado: Ejecutado; conservar las tres franjas provisionalmente.


### 3.4. Brillo y contraste

Concepto: Independiente; fotometría

Variable/dimensión: Brillo y contraste de entrada

Definición operativa: Variación determinista de brillo o contraste en el TIFF de validación antes del preprocesamiento normal.

Valores / escala: Multiplicadores 0.95 y 1.05; referencia 1.00.

Indicador: ΔMacro-F1, recall y precisión RUSSET, cambios de etiqueta y cambio absoluto medio p(RUSSET).

Fuente de datos: 6. entrenamiento/04_Importancia_Variables.ipynb; sensibilidad_fotometrica_validation.csv.

Estado: Ejecutado; no cambiar el preprocesamiento productivo por este ensayo.


### 3.5. Interpretación con Grad-CAM

Concepto: Apoyo interpretativo

Variable/dimensión: Regiones activadas: Grad-CAM

Definición operativa: Visualización cualitativa de regiones con mayor influencia en la activación de la clase analizada; no es una métrica causal ni una máscara de defecto.

Valores / escala: Mapas de calor superpuestos para ejemplos de validación.

Indicador: Ejemplos y regiones visualizadas; sin puntuación cuantitativa de localización.

Fuente de datos: gradcam_ejemplos_validation.png.

Estado: Ejecutado como complemento cualitativo.


### 3.6. Unidad de análisis

Concepto: Unidad de análisis

Variable/dimensión: Archivo TIFF de un fruto

Definición operativa: Se usa el TIFF completo de una palta Hass con tres franjas; sus vistas no se separan en particiones diferentes.

Valores / escala: 829 originales, 4 exclusiones lógicas, 825 elegibles: 572 train, 127 validation y 126 TEST.

Indicador: Conteo de TIFF por partición, clases e integridad SHA-256.

Fuente de datos: 4. EDA/01_EDA.ipynb y manifiestos de auditoría.

Estado: Verificado; la independencia entre frutos/lotes requiere trazabilidad adicional.

## 5. Alcance y precauciones metodológicas

- Macro-F1 histórico FT_D1_PHOTO en validation: 0.7876. En su reproducción posterior: 0.7781, con 126/127 etiquetas iguales al historial y diferencia máxima de probabilidad 0.004967. No declarar reproducción exacta.
- La hipótesis general y la del primer objetivo resumen contrastes exploratorios entre configuraciones; no se identificó por separado un efecto causal de cada cambio de entrenamiento.
- Las hipótesis se han reformulado a partir de observaciones del docente cuando los resultados de la semana 7 ya existían. No deben presentarse como hipótesis preregistradas antes de las pruebas.
- Una mejora aislada al ocultar la cámara 2 no demuestra que pueda eliminarse: el intervalo bootstrap incluye cero y el desenfoque de esa misma franja produce deterioro.
- Las transformaciones de ocultación y desenfoque alteran la distribución de las imágenes; la sensibilidad observada no determina calidad física ni superioridad de una cámara.
- Las imágenes TIFF se tratan como unidades de evaluación. La auditoría de duplicados exactos no acredita por sí sola independencia estadística entre frutos, lotes o capturas relacionadas.
- TEST se utilizó una vez para el modelo FT_D1_PHOTO previamente seleccionado. No utilizar esas métricas para ajustar hiperparámetros o seleccionar una nueva configuración; cualquier modelo futuro requerirá una evaluación independiente apropiada.
